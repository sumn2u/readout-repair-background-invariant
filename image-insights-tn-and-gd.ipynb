```python
# ============================================================
# Image statistics: GD vs TrashNet
# - skips *_mask.png (and other mask-like files)
# - computes statistics over ALL non-mask images (no subsampling)
# - GD rank-sum grid shows RAW images (no color tint)
# - reports mean entropy + mean FSR per dataset
# - plots sample images for both datasets
# ============================================================

import os
import numpy as np
import pandas as pd
import cv2
from PIL import Image
from tqdm.auto import tqdm
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. Robust dataset discovery
# ============================================================

IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
GD_CLASSES = ["trash", "metal", "paper", "glass", "plastic", "cardboard"]
TN_CLASSES = ["cardboard", "glass", "metal", "paper", "plastic", "trash"]


def find_dataset(candidate_roots, required_class, extra_search=None):
    for root in candidate_roots:
        if not os.path.isdir(root):
            continue
        if os.path.isdir(os.path.join(root, required_class)):
            return root
        for entry in os.listdir(root):
            sub = os.path.join(root, entry)
            if not os.path.isdir(sub):
                continue
            if os.path.isdir(os.path.join(sub, required_class)):
                return sub
            for entry2 in os.listdir(sub):
                sub2 = os.path.join(sub, entry2)
                if os.path.isdir(sub2) and os.path.isdir(
                        os.path.join(sub2, required_class)):
                    return sub2
    if extra_search:
        for root in extra_search:
            if os.path.isdir(root) and os.path.isdir(
                    os.path.join(root, required_class)):
                return root
    return None


GD_PATH = find_dataset(
    [
        "/kaggle/input/garbage-classification-v2/standardized_384",
        "/kaggle/input/garbage-classification-v2",
        "/kaggle/input/datasets/sumn2u/garbage-classification-v2",
        "/kaggle/input/datasets/sumn2u/garbage-classification-v2/standardized_384",
    ],
    required_class="trash",
)

TN_PATH = find_dataset(
    [
        "/kaggle/input/trashnet",
        "/kaggle/input/trashnet/dataset-resized",
        "/kaggle/input/datasets/sumn2u/trashnet",
        "/kaggle/input/datasets/sumn2u/trashnet/dataset-resized",
    ],
    required_class="cardboard",
)

print("GD path:", GD_PATH)
print("TrashNet path:", TN_PATH)

DATASETS = {}
if GD_PATH:
    DATASETS["GD"] = {"path": GD_PATH, "classes": GD_CLASSES}
if TN_PATH:
    DATASETS["TrashNet"] = {"path": TN_PATH, "classes": TN_CLASSES}

print("\nDatasets:", list(DATASETS.keys()))

if not DATASETS:
    raise RuntimeError("No datasets found. Attach them via 'Add Data'.")


# ============================================================
# 2. Per-image statistics
# ============================================================

def compute_stats(img_path, mask=None):
    img = cv2.imread(img_path, cv2.IMREAD_COLOR)
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, (384, 384))

    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY).astype(np.float32)
    hsv = cv2.cvtColor(img, cv2.COLOR_RGB2HSV).astype(np.float32)
    sat = hsv[..., 1] / 255.0
    val = hsv[..., 2] / 255.0

    stats = {
        "brightness": float(gray.mean() / 255.0),
        "luminance_std": float(gray.std() / 255.0),
        "saturation": float(sat.mean()),
        "value_mean": float(val.mean()),
    }

    hist = cv2.calcHist([gray.astype(np.uint8)], [0], None,
                        [256], [0, 256]).ravel()
    p = hist / (hist.sum() + 1e-12)
    p = p[p > 0]
    stats["entropy"] = float(-(p * np.log2(p)).sum())

    lap = cv2.Laplacian(gray, cv2.CV_32F)
    stats["laplacian_var"] = float(lap.var())

    hist_mode = int(np.argmax(hist))
    deviation = np.abs(gray - hist_mode)
    threshold = 1.0 * gray.std() + 1e-6
    stats["fsr"] = float((deviation > threshold).mean())

    if mask is not None:
        mask = cv2.resize(mask, (384, 384))
        m = (mask > 0.5)
        if m.sum() > 100 and (~m).sum() > 100:
            stats["fg_bg_luminance_diff"] = float(
                abs(gray[m].mean() - gray[~m].mean()) / 255.0)
            fg_rgb = img[m].astype(np.float32)
            bg_rgb = img[~m].astype(np.float32)
            stats["fg_bg_rgb_diff"] = float(
                np.linalg.norm(fg_rgb.mean(axis=0) - bg_rgb.mean(axis=0)) / 255.0)
        else:
            stats["fg_bg_luminance_diff"] = np.nan
            stats["fg_bg_rgb_diff"] = np.nan

    return stats


def load_rgb_safe(path, size=None):
    img = cv2.imread(path, cv2.IMREAD_COLOR)
    if img is None:
        placeholder = np.full((224, 224, 3), 220, dtype=np.uint8)
        cv2.putText(placeholder, "LOAD FAIL", (30, 110),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.7, (200, 0, 0), 2)
        return placeholder
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    if size is not None:
        img = cv2.resize(img, size)
    return img


# ============================================================
# 2b. Robust mask-file detection
#     Catches: *_mask.png, mask_*.png, *_seg.png, and
#     near-binary PNG/BMP files that are almost certainly masks.
# ============================================================

def is_mask_file(fname, cls_path):
    low = fname.lower()
    # explicit mask markers in the filename
    if "_mask" in low or "mask_" in low or "_seg" in low or "segmentation" in low:
        return True

    # heuristic: PNG / BMP that is grayscale or palette with <=4 unique values
    ext = Path(fname).suffix.lower()
    if ext in {".png", ".bmp"}:
        p = os.path.join(cls_path, fname)
        try:
            im = Image.open(p)
            if im.mode in ("L", "1", "P"):
                arr = np.array(im.convert("L"))
                if len(np.unique(arr)) <= 4:   # near-binary -> mask
                    return True
        except Exception:
            # unreadable file -> treat as non-image, skip anyway
            return True
    return False


# ============================================================
# 3. Enumerate + compute stats  (SKIPS masks robustly)
#    [FULL-DATASET] all non-mask images are used; no subsampling.
# ============================================================

all_stats = []
skipped_masks = 0

for ds_name, cfg in DATASETS.items():
    print(f"\n--- {ds_name} ---")
    root = cfg["path"]
    records = []
    for cls in cfg["classes"]:
        cls_path = os.path.join(root, cls)
        if not os.path.isdir(cls_path):
            continue
        for f in sorted(os.listdir(cls_path)):
            if Path(f).suffix.lower() not in IMG_EXTS:
                continue
            if is_mask_file(f, cls_path):     # <-- robust skip
                skipped_masks += 1
                continue
            records.append({"path": os.path.join(cls_path, f),
                            "class": cls})
    print(f"  Total images: {len(records)}  "
          f"(skipped {skipped_masks} masks so far)")

    # [FULL-DATASET] use every non-mask image (no per-class subsampling)
    sampled = records
    print(f"  Using all {len(sampled)} images for statistics")

    for r in tqdm(sampled, desc=ds_name):
        s = compute_stats(r["path"])
        if s is None:
            continue
        s["dataset"] = ds_name
        s["class"] = r["class"]
        s["path"] = r["path"]
        all_stats.append(s)

print(f"\nTotal mask-like files skipped across all datasets: {skipped_masks}")
stats_df = pd.DataFrame(all_stats)
print(f"Total rows collected: {len(stats_df)}")

if len(stats_df) == 0:
    raise RuntimeError("No statistics were collected.")


# ============================================================
# 3b. Audit: verify no mask files leaked into stats_df
# ============================================================

mask_like = stats_df["path"].str.lower().str.contains(
    "_mask|mask_|_seg|segmentation", regex=True
)
print(f"\nMask-like paths remaining in stats_df: {mask_like.sum()}")
if mask_like.sum() > 0:
    print("  (should be 0 — check the is_mask_file heuristic)")

print("\nFile extensions per dataset:")
for ds in stats_df["dataset"].unique():
    exts = (
        stats_df[stats_df["dataset"] == ds]["path"]
        .map(lambda p: Path(p).suffix.lower())
        .value_counts()
    )
    print(f"  {ds}: {exts.to_dict()}")


# ============================================================
# 4. Summary
# ============================================================

METRICS = ["brightness", "luminance_std", "saturation",
           "value_mean", "entropy", "laplacian_var", "fsr"]

print("\n" + "=" * 70)
print("PER-DATASET SUMMARY")
print("=" * 70)
print(stats_df.groupby("dataset")[METRICS].mean().T.to_string())

print("\n" + "=" * 70)
print("PER-CLASS SUMMARY")
print("=" * 70)
print(stats_df.groupby(["dataset", "class"])[METRICS].mean().to_string())


# ============================================================
# 4b. Mean entropy and mean FSR per dataset
# ============================================================

print("\n" + "=" * 70)
print("MEAN ENTROPY and MEAN FSR PER DATASET")
print("=" * 70)

mean_report = (
    stats_df.groupby("dataset")[["entropy", "fsr"]]
            .agg(["mean", "std", "median"])
            .round(4)
)
print(mean_report.to_string())

print("\n" + "=" * 70)
print("MEAN ENTROPY and MEAN FSR PER (dataset, class)")
print("=" * 70)
per_class = (
    stats_df.groupby(["dataset", "class"])[["entropy", "fsr"]]
            .agg(["mean", "std", "median"])
            .round(4)
)
print(per_class.to_string())

mean_report.to_csv("/kaggle/working/mean_entropy_fsr_by_dataset.csv")
per_class.to_csv("/kaggle/working/mean_entropy_fsr_by_class.csv")
print("\nSaved mean entropy/FSR tables to /kaggle/working/")


# ============================================================
# 4c. Sample grids per dataset  (illustrations only)
# ============================================================

def plot_dataset_samples(df, dataset_name, per_class=1, tag=""):
    sub = df[df["dataset"] == dataset_name]
    if len(sub) == 0:
        print(f"No rows for {dataset_name}; skipping sample grid.")
        return

    picks = (
        sub.groupby("class", group_keys=False)
           .apply(lambda g: g.sample(min(per_class, len(g)), random_state=42))
           .reset_index(drop=True)
    )

    n = len(picks)
    cols = min(4, n)
    rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 3.4, rows * 3.4))
    axes = np.array(axes).reshape(-1)

    for ax, (_, r) in zip(axes, picks.iterrows()):
        img = load_rgb_safe(r["path"])
        ax.imshow(img)
        ax.set_title(
            f"{r['class']}\nH={r['entropy']:.2f}  FSR={r['fsr']:.2f}",
            fontsize=9,
        )
        ax.axis("off")

    for ax in axes[n:]:
        ax.axis("off")

    plt.suptitle(f"{dataset_name} samples {tag}", fontsize=12)
    plt.tight_layout()
    safe_tag = tag.replace(" ", "_").replace("(", "").replace(")", "")
    out = f"/kaggle/working/{dataset_name.lower()}_samples{safe_tag}.png"
    plt.savefig(out, dpi=140)
    plt.show()
    print(f"Saved {out}")


# One sample per class for each dataset
plot_dataset_samples(stats_df, "GD", per_class=1, tag="(1 per class)")
plot_dataset_samples(stats_df, "TrashNet", per_class=1,
                     tag="(1 per class, masks excluded)")

# Two samples per class for a richer grid
plot_dataset_samples(stats_df, "GD", per_class=2, tag="(2 per class)")
plot_dataset_samples(stats_df, "TrashNet", per_class=2,
                     tag="(2 per class, masks excluded)")


# ============================================================
# 5. Box-plot visualizations
# ============================================================

fig, axes = plt.subplots(2, 3, figsize=(18, 10))
for ax, m in zip(axes.flat, ["brightness", "luminance_std", "saturation",
                             "entropy", "laplacian_var", "fsr"]):
    sns.boxplot(data=stats_df, x="dataset", y=m, ax=ax)
    ax.set_title(m)
    ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig("/kaggle/working/dataset_stats.png", dpi=120)
plt.show()
plt.close()

stats_df.to_csv("/kaggle/working/dataset_stats.csv", index=False)
print("\nSaved stats to /kaggle/working/dataset_stats.csv")


# ============================================================
# 6. GD samples: HIGHEST entropy + LOWEST FSR (rank-sum blend)
#    NOTE: raw images, no red overlay / no color tint.
# ============================================================

gd = stats_df[stats_df["dataset"] == "GD"].copy()

if len(gd) == 0:
    print("No GD rows found; skipping sample grid.")
else:
    gd["rank_entropy"] = gd["entropy"].rank(ascending=False, method="min")
    gd["rank_fsr"]     = gd["fsr"].rank(ascending=True,  method="min")
    gd["rank_sum"]     = gd["rank_entropy"] + gd["rank_fsr"]

    top = (
        gd.sort_values("rank_sum")
          .groupby("class", group_keys=False)
          .head(2)
          .sort_values("rank_sum")
          .head(12)
          .reset_index(drop=True)
    )

    print("\nTop GD matches (highest entropy, lowest FSR — rank-sum blend):")
    print(top[["class", "entropy", "fsr",
               "rank_entropy", "rank_fsr", "rank_sum"]].to_string(index=False))

    n = len(top)
    cols = 4
    rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 3.4, rows * 3.4))
    axes = np.array(axes).reshape(-1)

    for ax, (_, row) in zip(axes, top.iterrows()):
        img = load_rgb_safe(row["path"])
        ax.imshow(img)                      # <-- raw image, no tint
        ax.set_title(
            f"{row['class']}\nH={row['entropy']:.2f}  FSR={row['fsr']:.3f}",
            fontsize=9,
        )
        ax.axis("off")

    for ax in axes[n:]:
        ax.axis("off")

    plt.suptitle("GD: highest background entropy + lowest FSR "
                 "(rank-sum blend)", fontsize=12)
    plt.tight_layout()
    plt.savefig("/kaggle/working/gd_top_entropy_low_fsr.png", dpi=140)
    plt.show()


# ============================================================
# 6b. Strict two-key sort (entropy DESC, then FSR ASC)
# ============================================================

if len(gd) > 0:
    top_strict = (
        gd.sort_values(["entropy", "fsr"], ascending=[False, True])
          .groupby("class", group_keys=False)
          .head(2)
          .sort_values(["entropy", "fsr"], ascending=[False, True])
          .head(12)
          .reset_index(drop=True)
    )

    print("\nTop GD matches (strict two-key sort):")
    print(top_strict[["class", "entropy", "fsr"]].to_string(index=False))

    n = len(top_strict)
    cols = 4
    rows = int(np.ceil(n / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 3.4, rows * 3.4))
    axes = np.array(axes).reshape(-1)

    for ax, (_, row) in zip(axes, top_strict.iterrows()):
        img = load_rgb_safe(row["path"])
        ax.imshow(img)
        ax.set_title(
            f"{row['class']}\nH={row['entropy']:.2f}  FSR={row['fsr']:.3f}",
            fontsize=9,
        )
        ax.axis("off")

    for ax in axes[n:]:
        ax.axis("off")

    plt.suptitle("GD: highest entropy, then lowest FSR (strict sort)",
                 fontsize=12)
    plt.tight_layout()
    plt.savefig("/kaggle/working/gd_top_entropy_low_fsr_strict.png", dpi=140)
    plt.show()


# ============================================================
# 7. Side-by-side: GD (high entropy) vs TrashNet (non-blank)
# ============================================================

if "GD" in stats_df["dataset"].unique() and "TrashNet" in stats_df["dataset"].unique():

    gd_sub = stats_df[stats_df["dataset"] == "GD"].copy()
    gd_picks = gd_sub.sort_values("entropy", ascending=False).head(6)

    tn_sub = stats_df[stats_df["dataset"] == "TrashNet"].copy()

    blank_mask = (tn_sub["laplacian_var"] < 50) | (tn_sub["luminance_std"] < 0.05)
    print(f"\nTrashNet: {blank_mask.sum()} / {len(tn_sub)} "
          f"images look near-blank (Lap<50 or lum_std<0.05)")

    tn_filtered = tn_sub[
        (tn_sub["laplacian_var"] > 100) &
        (tn_sub["luminance_std"] > 0.08)
    ].copy()

    if len(tn_filtered) == 0:
        print("WARNING: all TrashNet images filtered as blank. "
              "Falling back to median-entropy picks.")
        med = tn_sub["entropy"].median()
        tn_picks = (
            tn_sub.assign(dist=(tn_sub["entropy"] - med).abs())
                  .sort_values("dist")
                  .groupby("class", group_keys=False)
                  .head(1)
                  .head(6)
        )
    else:
        tn_picks = tn_filtered.sort_values(
            "laplacian_var", ascending=False
        ).head(6)

    fig, axes = plt.subplots(2, 6, figsize=(20, 7))
    for row_idx, (label, picks) in enumerate(
        [("GD (highest entropy)", gd_picks),
         ("TrashNet (most textured)", tn_picks)]
    ):
        for col_idx, (_, r) in enumerate(picks.iterrows()):
            ax = axes[row_idx, col_idx]
            img = load_rgb_safe(r["path"])
            ax.imshow(img)
            ax.set_title(
                f"{r['class']}\n"
                f"H={r['entropy']:.2f}  FSR={r['fsr']:.2f}  "
                f"Lap={r['laplacian_var']:.0f}",
                fontsize=8,
            )
            ax.axis("off")

    plt.suptitle(
        "GD highest-entropy (top) vs TrashNet most-textured (bottom)",
        fontsize=13,
    )
    plt.tight_layout()
    plt.savefig("/kaggle/working/gd_vs_trashnet_samples.png", dpi=140)
    plt.show()

    print("\nTrashNet picks (most-textured):")
    for _, r in tn_picks.iterrows():
        print(f"  {r['class']:10s} H={r['entropy']:.2f} "
              f"Lap={r['laplacian_var']:.0f}  {r['path']}")

    # ---------- 7b. Lowest-entropy TrashNet picks ----------
    blank_samples = (
        tn_sub.sort_values("entropy")
              .head(6)
              .reset_index(drop=True)
    )

    fig, axes = plt.subplots(1, 6, figsize=(20, 4))
    for ax, (_, r) in zip(axes, blank_samples.iterrows()):
        img = load_rgb_safe(r["path"])
        ax.imshow(img)
        ax.set_title(
            f"{r['class']}\nH={r['entropy']:.2f}  Lap={r['laplacian_var']:.0f}",
            fontsize=8,
        )
        ax.axis("off")
    plt.suptitle("TrashNet lowest-entropy samples "
                 "(near-uniform, low texture)", fontsize=11)
    plt.tight_layout()
    plt.savefig("/kaggle/working/trashnet_lowest_entropy.png", dpi=140)
    plt.show()

    print("\n" + "=" * 90)
    print("TRASHNET LOWEST-ENTROPY PICKS")
    print("=" * 90)
    for i, r in blank_samples.iterrows():
        print(f"\n[{i+1}] class={r['class']}")
        print(f"    entropy        = {r['entropy']:.3f} bits")
        print(f"    fsr            = {r['fsr']:.3f}")
        print(f"    laplacian_var  = {r['laplacian_var']:.1f}")
        print(f"    luminance_std  = {r['luminance_std']:.3f}")
        print(f"    path           = {r['path']}")

else:
    print("Skipping GD vs TrashNet comparison — one dataset missing.")
```
